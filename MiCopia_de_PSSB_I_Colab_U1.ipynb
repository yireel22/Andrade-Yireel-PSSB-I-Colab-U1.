- *¿Cuáles son las fases de la etapa de Análisis (Front-End) de un compilador?*  
**RESPUESTA:**

La etapa de **Análisis (también conocida como Front-End)** de un compilador se enfoca en entender el programa fuente, verificar que sea correcto tanto en su estructura como en su significado, y registrar los resultados de este análisis para preparar la posterior síntesis del código. Tradicionalmente, esta etapa comienza con el código fuente y finaliza con la creación de una representación intermedia del programa.

Para llevar a cabo su tarea, el Front-End se divide en las siguientes fases principales:

### 1. Análisis Léxico (o Lexicográfico / Morfológico / Escaneo)
Es la primera fase del compilador. Su función consiste en leer el flujo de caracteres del código fuente de izquierda a derecha para agruparlos en secuencias lógicas llamadas **lexemas**. Para cada lexema, el analizador léxico genera una entidad atómica llamada **token** (o componente léxico), la cual consta de una categoría (su tipo) y un atributo (su valor o lexema).
*   **Tareas secundarias:** Se encarga de eliminar comentarios, espacios en blanco, tabulaciones y saltos de línea que carezcan de significado para la sintaxis del lenguaje. También lleva la cuenta del número de línea para poder asociar y localizar los mensajes de error en el programa fuente.

### 2. Análisis Sintáctico (o *Parsing*)
Toma como entrada la secuencia de tokens suministrada por el analizador léxico y comprueba que estén ordenados de forma correcta según las reglas gramaticales del lenguaje (definidas mediante una gramática independiente del contexto).
*   **Representación interna:** Conceptualmente, si la secuencia es correcta, esta fase construye una estructura jerárquica en forma de **árbol sintáctico**. En este árbol, los nodos interiores representan las operaciones y las hojas representan los operandos. El analizador sintáctico es típicamente el que guía todo el proceso de compilación (compilación dirigida por la sintaxis) solicitando activamente tokens al analizador léxico y llamando a las rutinas de las siguientes fases.

### 3. Análisis Semántico
Esta fase revisa el árbol sintáctico teórico y la tabla de símbolos para verificar que las instrucciones tengan un significado coherente y respeten las directrices semánticas del lenguaje.
*   **Verificación de tipos:** Es su componente más importante. El analizador comprueba que los operandos que intervienen en cada operador sean compatibles (por ejemplo, evitar que se sume un entero con una cadena de caracteres o que se indexe un arreglo usando un número flotante).
*   **Otras comprobaciones:** Verifica que las variables hayan sido declaradas antes de ser utilizadas, que las llamadas a funciones tengan el número y tipo de parámetros correctos, y gestiona los ámbitos o regiones de alcance de los identificadores. El resultado de esta fase es un árbol de derivación con anotaciones semánticas.

### 4. Generación de Código Intermedio
Aunque es una fase de transición hacia la etapa de síntesis, comúnmente se integra dentro de las fases del Front-End. A partir del árbol semántico anotado, genera una representación explícita y de bajo nivel del programa fuente para una **máquina abstracta virtual**.
*   **Propiedades:** Este código abstracto (como el código de tres direcciones o cuádruplas) debe ser fácil de producir y fácil de traducir al lenguaje máquina destino de cualquier procesador. Su uso permite independizar el análisis del lenguaje fuente de los detalles específicos del hardware final.

### Componentes de soporte en todo el Front-End

Durante todas las fases de análisis, el compilador se apoya de manera intensiva en dos estructuras transversales:

*   **La Tabla de Símbolos:** Es una estructura de datos (como una tabla hash o árbol) que funciona como un diccionario de datos. Almacena un registro para cada identificador detectado en el programa fuente junto con sus atributos (tipo, espacio en memoria, ámbito, etc.). Todas las fases del Front-End acceden a ella para insertar información en las declaraciones o consultar datos al momento de utilizar las variables.
*   **El Manejador de Errores:** Cada fase de análisis debe ser capaz de detectar errores, reportar mensajes claros indicando la línea donde se produjeron, e implementar estrategias de recuperación (como el modo de pánico o la recuperación a nivel de frase) para poder continuar analizando el resto del programa sin detener la compilación por completo.





- *¿Qué diferencia a un Lexema de un Token?*  
**RESPUESTA:**
La diferencia principal entre un lexema y un token radica en el nivel de abstracción y en el papel que desempeñan dentro del proceso de compilación:
Lexema: Es la secuencia concreta de caracteres en el programa fuente que coincide con el patrón definido para un componente léxico
. Representa la cadena de texto original tal y como fue escrita por el programador (por ejemplo, el nombre de una variable específica como "cuenta" o un número determinado como "31")
. Es, en esencia, la instancia física o el valor real que encaja dentro de una regla
.
Token: Es la categoría léxica abstracta o el nombre lógico asignado a un patrón definido
. Funciona como una unidad lógica indivisible que el analizador léxico entrega como entrada al analizador sintáctico, donde equivale directamente al concepto de símbolo terminal en la gramática
. El token suele constar de dos componentes esenciales: el nombre del token (o categoría) y un valor de atributo opcional que proporciona detalles semánticos (como un puntero a la tabla de símbolos o el valor numérico correspondiente)
.
Relación e Interacción entre Ambos:
No existe una relación uno a uno: Un solo token de categoría general (como un identificador ID o un número NUM) puede representar e identificarse con infinitos lexemas distintos en el código fuente
. Por el contrario, los tokens asociados a palabras reservadas (como while o if) suelen corresponder a un único lexema específico
.
La analogía de las frutas: El token equivale a una categoría general (como la palabra "fruta"), mientras que el lexema representa cada elemento físico y concreto que encaja en ella (como "manzana", "plátano", etc.)
.
Aislamiento del compilador: El analizador léxico traduce los lexemas en tokens para aislar al analizador sintáctico de las variaciones textuales y los caracteres del código fuente
. Para lograr esto, el analizador léxico suministra la información al sintáctico en forma de un par o pareja de datos representados como (token, lexema) o 〈nombre-token, valor-atributo〉

## El "Monstruo Completo" (Enlace Estático)  
Por defecto, GCC usa ***enlace dinámico*** usando una estructura de datos tabular, la tabla **PLT** (*Procedure Linkage Table*) ya que el código usa `printf`, el programa necesita conectarse con la biblioteca estándar de C (`libc.so`). El enlazador no copia el código de `printf` dentro del archivo para no duplicar espacio en el disco; en su lugar, crea un "trampolín" o acceso directo en una sección llamada `<printf@plt>`.

Cada vez que se invoca a `printf`, realmente salta a este bloque plt, el cual averigua en qué parte de la memoria RAM del sistema operativo se encuentra la verdadera función `printf` y redirige el control allí en tiempo de ejecución.

Para ver cómo el enlazador inyecta literalmente miles y miles de líneas copiando físicamente el código de `printf` y todas sus dependencias dentro del ejecutable, se debe compilar de forma estática:

````python
import token
import tokenize
from io import BytesIO

# Código fuente de prueba en formato cadena
codigo_fuente = "suma = a + 10"

# Tokenización del código fuente
tokens = tokenize.tokenize(BytesIO(codigo_fuente.encode("utf-8")).readline)

print(
    f"{'LINEA/COL':<12} | {'TIPO DE TOKEN':<20} | {'VALOR (LEXEMA)':<15}"
)
print("-" * 55)
for tok in tokens:
    if tok.type in (
        tokenize.ENCODING,
        tokenize.ENDMARKER,
        tokenize.NL,
        tokenize.NEWLINE,
    ):
        continue
    nombre_token = token.tok_name[tok.type]
    posicion = f"{tok.start[0]}:{tok.start[1]}"
    print(f"{posicion:<12} | {nombre_token:<20} | {tok.string:<15}")
````

            self.simbolos[nombre] = {

2. Implementa una función de prueba preliminar (o utiliza la librería `tokenize` de Python como simulador) que tome la cadena de tu lenguaje y la desglose en una lista de componentes léxicos.

````python
import tokenize
import io
import ast
from typing import List, Dict, Any, Tuple
from enum import Enum
from dataclasses import dataclass
from collections import Counter
import re

# ==========================================
# 1. FUNCIÓN DE PRUEBA CON TOKENIZE
# ==========================================

def analizar_con_tokenize(codigo: str) -> List[Dict[str, Any]]:
    """
    Utiliza la librería tokenize de Python para desglosar el código
    en componentes léxicos (tokens)

    Args:
        codigo: Cadena con el código fuente

    Returns:
        Lista de diccionarios con información de cada token
    """
    tokens = []

    # Crear un buffer de lectura para tokenize
    buffer = io.StringIO(codigo)

    try:
        # Generar tokens usando tokenize
        for token in tokenize.generate_tokens(buffer.readline):
            # token es una tupla: (tipo, string, inicio, fin, linea)
            token_info = {
                'tipo': tokenize.tok_name[token.type],
                'tipo_numero': token.type,
                'valor': token.string,
                'linea_inicio': token.start[0],
                'columna_inicio': token.start[1],
                'linea_fin': token.end[0],
                'columna_fin': token.end[1],
                'linea_completa': token.line.strip() if token.line else ''
            }
            tokens.append(token_info)
    except tokenize.TokenError as e:
        print(f"❌ Error al tokenizar: {e}")

    return tokens

def mostrar_tokens_formateados(tokens: List[Dict[str, Any]], max_tokens: int = None):
    """
    Muestra los tokens en formato tabla legible

    Args:
        tokens: Lista de tokens generados por analizar_con_tokenize
        max_tokens: Número máximo de tokens a mostrar (None = todos)
    """
    print("\n" + "="*100)
    print("🔍 ANÁLISIS LÉXICO CON TOKENIZE DE PYTHON")
    print("="*100)

    # Mostrar estadísticas generales
    print(f"\n📊 Estadísticas:")
    print(f"   • Total de tokens: {len(tokens)}")

    # Contar tipos de tokens
    tipos = Counter([t['tipo'] for t in tokens])
    print(f"   • Tipos de tokens únicos: {len(tipos)}")
    print(f"   • Tipos más comunes:")
    for tipo, count in tipos.most_common(5):
        print(f"       - {tipo}: {count} ({count/len(tokens)*100:.1f}%)")

    # Mostrar tabla de tokens
    print("\n📋 TABLA DE TOKENS:")
    print("-"*100)
    print(f"{'#':<5} {'TIPO':<25} {'VALOR':<20} {'LÍNEA':<8} {'COLUMNA':<10} {'CONTEXTO':<15}")
    print("-"*100)

    # Limitar número de tokens a mostrar si se especifica
    tokens_mostrar = tokens if max_tokens is None else tokens[:max_tokens]

    for i, token in enumerate(tokens_mostrar):
        # Resaltar tokens especiales
        valor = token['valor']
        if token['tipo'] == 'NAME' and token['valor'] in ['def', 'if', 'for', 'while', 'return']:
            valor = f"🔑 {valor}"  # Palabras reservadas
        elif token['tipo'] == 'STRING':
            valor = f"📝 {valor}"  # Cadenas
        elif token['tipo'] == 'NUMBER':
            valor = f"🔢 {valor}"  # Números
        elif token['tipo'] == 'OP':
            valor = f"⚙️  {valor}"  # Operadores

        # Determinar contexto
        contexto = "Código"
        if token['tipo'] == 'COMMENT':
            contexto = "Comentario"
        elif token['tipo'] == 'NL' or token['tipo'] == 'NEWLINE':
            contexto = "Nueva línea"
        elif token['tipo'] == 'INDENT':
            contexto = "Indentación"
        elif token['tipo'] == 'DEDENT':
            contexto = "Desindentación"

        print(f"{i+1:<5} {token['tipo']:<25} {valor:<20} {token['linea_inicio']:<8} "
              f"({token['columna_inicio']:<3}-{token['columna_fin']:<3})   {contexto:<15}")

    if max_tokens and len(tokens) > max_tokens:
        print(f"\n... y {len(tokens) - max_tokens} tokens más")

    print("-"*100)


# ==========================================
# 2. ANÁLISIS DE COMPONENTES LÉXICOS PERSONALIZADO
# ==========================================

class ComponenteLexico:
    """Clase que representa un componente léxico del lenguaje Python"""

    def __init__(self, tipo: str, valor: str, linea: int, columna: int):
        self.tipo = tipo
        self.valor = valor
        self.linea = linea
        self.columna = columna

    def __str__(self):
        return f"{self.tipo:15} | '{self.valor:20}' | L{self.linea:3} C{self.columna:3}"

def desglosar_componentes_lexicos(codigo: str) -> List[ComponenteLexico]:
    """
    Desglosa el código en componentes léxicos usando análisis manual
    (complementario a tokenize)

    Args:
        codigo: Cadena con el código fuente

    Returns:
        Lista de objetos ComponenteLexico
    """
    componentes = []
    linea_actual = 1
    columna_actual = 1
    i = 0

    # Patrones para reconocer componentes léxicos
    patrones = [
        ('PALABRA_RESERVADA', r'^(def|if|else|elif|for|while|return|import|from|class|try|except|finally|with|as|pass|break|continue|lambda|yield|global|nonlocal|True|False|None|and|or|not|in|is)'),
        ('IDENTIFICADOR', r'^[a-zA-Z_][a-zA-Z0-9_]*'),
        ('NUMERO', r'^\d+\.\d+|^\d+'),
        ('CADENA', r'^".*?"|^\'.*?\''),
        ('OPERADOR', r'^[+\-*/%=<>!&|^~]+'),
        ('SIMBOLO', r'^[()\[\]{};,.:]'),
        ('COMENTARIO', r'^#.*'),
        ('ESPACIO', r'^\s+'),
    ]

    while i < len(codigo):
        caracter = codigo[i]

        # Saltar espacios
        if caracter.isspace():
            if caracter == '\n':
                linea_actual += 1
                columna_actual = 1
            else:
                columna_actual += 1
            i += 1
            continue

        # Buscar coincidencia con patrones
        coincidencia = False
        for tipo, patron in patrones:
            # Usar re.match para buscar al inicio de la posición actual
            match = re.match(patron, codigo[i:])
            if match:
                valor = match.group(0)
                # Solo procesar si no es espacio
                if tipo != 'ESPACIO':
                    componentes.append(ComponenteLexico(tipo, valor, linea_actual, columna_actual))
                i += len(valor)
                columna_actual += len(valor)
                coincidencia = True
                break

        # Si no coincide con ningún patrón, es un carácter desconocido
        if not coincidencia:
            print(f"⚠️  Carácter desconocido: '{caracter}' en línea {linea_actual}, columna {columna_actual}")
            i += 1
            columna_actual += 1

    return componentes

def mostrar_componentes_lexicos(componentes: List[ComponenteLexico]):
    """
    Muestra los componentes léxicos en formato tabla
    """
    print("\n" + "="*100)
    print("📝 COMPONENTES LÉXICOS (Análisis Manual)")
    print("="*100)

    print(f"\n📊 Total de componentes: {len(componentes)}")

    # Clasificar por tipo
    tipos = Counter([c.tipo for c in componentes])
    print(f"\n📊 Distribución por tipo:")
    for tipo, count in sorted(tipos.items(), key=lambda x: x[1], reverse=True):
        print(f"   • {tipo}: {count} ({count/len(componentes)*100:.1f}%)")

    print("\n📋 LISTA DE COMPONENTES:")
    print("-"*70)
    print(f"{'TIPO':<15} {'VALOR':<25} {'UBICACIÓN'}")
    print("-"*70)

    for comp in componentes[:50]:  # Mostrar primeros 50
        print(comp)

    if len(componentes) > 50:
        print(f"... y {len(componentes) - 50} componentes más")

    print("-"*70)


# ==========================================
# 3. CLASE PARA INSPECCIÓN COMPLETA
# ==========================================

class InspectorLexicoPython:
    """
    Clase que integra todos los análisis léxicos
    """

    def __init__(self, codigo: str):
        self.codigo = codigo
        self.tokens_tokenize = None
        self.componentes_manuales = None

    def inspeccionar_completo(self):
        """Ejecuta todas las inspecciones léxicas"""
        print("\n" + "="*100)
        print("🔬 INSPECCIÓN LÉXICA COMPLETA - PYTHON")
        print("="*100)

        # 1. Análisis con tokenize
        print("\n📌 1. ANÁLISIS CON TOKENIZE (Librería estándar)")
        self.tokens_tokenize = analizar_con_tokenize(self.codigo)
        mostrar_tokens_formateados(self.tokens_tokenize, max_tokens=30)

        # 2. Análisis manual de componentes léxicos
        print("\n📌 2. ANÁLISIS MANUAL DE COMPONENTES LÉXICOS")
        self.componentes_manuales = desglosar_componentes_lexicos(self.codigo)
        mostrar_componentes_lexicos(self.componentes_manuales)

        # 3. Comparación entre ambos métodos
        self._comparar_analisis()

        # 4. Análisis detallado de tokens
        self._analizar_tokens_detallado()

    def _comparar_analisis(self):
        """Compara los resultados de ambos métodos de análisis"""
        print("\n📌 3. COMPARACIÓN DE MÉTODOS DE ANÁLISIS")
        print("-"*100)

        # Contar tokens de tokenize
        if self.tokens_tokenize:
            token_count = len(self.tokens_tokenize)
            tipos_tokenize = Counter([t['tipo'] for t in self.tokens_tokenize])
        else:
            token_count = 0
            tipos_tokenize = {}

        # Contar componentes manuales
        if self.componentes_manuales:
            comp_count = len(self.componentes_manuales)
            tipos_manual = Counter([c.tipo for c in self.componentes_manuales])
        else:
            comp_count = 0
            tipos_manual = {}

        print(f"\n   {token_count:<20} vs {comp_count:<20} (Total)")
        print(f"   {'TOKENIZE':<20} vs {'MANUAL':<20} (Método)")
        print(f"   {'-'*20}   vs {'-'*20}")

        # Comparar tipos comunes
        tipos_comunes = set(tipos_tokenize.keys()) & set(tipos_manual.keys())
        print(f"\n   Tipos coincidentes: {len(tipos_comunes)}")

        # Mostrar diferencias
        solo_tokenize = set(tipos_tokenize.keys()) - set(tipos_manual.keys())
        solo_manual = set(tipos_manual.keys()) - set(tipos_tokenize.keys())

        if solo_tokenize:
            print(f"\n   📌 Tipos solo en tokenize: {', '.join(solo_tokenize)}")
        if solo_manual:
            print(f"   📌 Tipos solo en manual: {', '.join(solo_manual)}")

    def _analizar_tokens_detallado(self):
        """Realiza un análisis detallado de los tokens"""
        print("\n📌 4. ANÁLISIS DETALLADO DE TOKENS")
        print("-"*100)

        if not self.tokens_tokenize:
            print("   No hay tokens para analizar")
            return

        # Agrupar por tipo
        tokens_por_tipo = {}
        for token in self.tokens_tokenize:
            tipo = token['tipo']
            if tipo not in tokens_por_tipo:
                tokens_por_tipo[tipo] = []
            tokens_por_tipo[tipo].append(token['valor'])

        # Mostrar agrupación
        print("\n   📊 Tokens agrupados por tipo:")
        for tipo, valores in sorted(tokens_por_tipo.items()):
            # Mostrar valores únicos y su frecuencia
            conteo = Counter(valores)
            print(f"\n   • {tipo}:")
            for valor, count in conteo.most_common(5):
                print(f"       - '{valor}' → {count} veces")

        # Identificar palabras reservadas
        palabras_reservadas = ['def', 'if', 'else', 'elif', 'for', 'while', 'return',
                              'import', 'from', 'class', 'try', 'except', 'finally',
                              'with', 'as', 'pass', 'break', 'continue', 'lambda',
                              'yield', 'global', 'nonlocal', 'True', 'False', 'None',
                              'and', 'or', 'not', 'in', 'is']

        reservadas_encontradas = [t for t in self.tokens_tokenize
                                 if t['tipo'] == 'NAME' and t['valor'] in palabras_reservadas]

        print(f"\n   🔑 Palabras reservadas encontradas: {len(reservadas_encontradas)}")
        reservadas_unicas = sorted(set([t['valor'] for t in reservadas_encontradas]))
        if reservadas_unicas:
            print(f"       {', '.join(reservadas_unicas)}")

        # Identificar identificadores
        identificadores = [t for t in self.tokens_tokenize
                          if t['tipo'] == 'NAME' and t['valor'] not in palabras_reservadas]

        print(f"\n   📌 Identificadores encontrados: {len(identificadores)}")
        identificadores_unicos = sorted(set([t['valor'] for t in identificadores]))
        if identificadores_unicos:
            print(f"       {', '.join(identificadores_unicos[:10])}")
            if len(identificadores_unicos) > 10:
                print(f"       ... y {len(identificadores_unicos) - 10} más")

        # Identificar literales (números y cadenas)
        numeros = [t for t in self.tokens_tokenize if t['tipo'] == 'NUMBER']
        cadenas = [t for t in self.tokens_tokenize if t['tipo'] == 'STRING']

        print(f"\n   🔢 Literales numéricos: {len(numeros)}")
        if numeros:
            print(f"       {', '.join(set([t['valor'] for t in numeros]))}")

        print(f"   📝 Literales de cadena: {len(cadenas)}")
        if cadenas:
            # Mostrar solo los primeros 5
            cadenas_mostrar = list(set([t['valor'][:30] + '...' if len(t['valor']) > 30 else t['valor']
                                       for t in cadenas]))[:5]
            print(f"       {', '.join(cadenas_mostrar)}")
````

````python
import ast
import inspect
from typing import Dict, List, Any, Optional, Set
from dataclasses import dataclass, field
from enum import Enum
from datetime import datetime
import hashlib

# ==========================================
# 1. DEFINICIONES DE LA TABLA DE SÍMBOLOS
# ==========================================

class TipoSimbolo(Enum):
    """Tipos de símbolos que puede almacenar la tabla"""
    VARIABLE = "variable"
    FUNCION = "function"
    CLASE = "class"
    PARAMETRO = "parametro"
    CONSTANTE = "constante"
    IMPORT = "import"
    MODULO = "modulo"

class TipoDato(Enum):
    """Tipos de datos soportados"""
    ENTERO = "int"
    FLOTANTE = "float"
    CADENA = "str"
    BOOLEANO = "bool"
    LISTA = "list"
    DICCIONARIO = "dict"
    TUPLA = "tuple"
    CONJUNTO = "set"
    NINGUNO = "None"
    DESCONOCIDO = "unknown"

class Ambito(Enum):
    """Ámbitos de visibilidad"""
    GLOBAL = "global"
    LOCAL = "local"
    NO_LOCAL = "nonlocal"
    BLOQUE = "block"
    CLASE = "class"
    FUNCION = "function"

@dataclass
class AtributosSimbolo:
    """Atributos completos de un símbolo en la tabla"""
    # Información básica
    nombre: str
    tipo_simbolo: TipoSimbolo
    tipo_dato: TipoDato = TipoDato.DESCONOCIDO

    # Información de valor
    valor: Any = None
    valor_inicial: Any = None
    es_constante: bool = False
    es_mutable: bool = True

    # Información de ámbito
    ambito: Ambito = Ambito.GLOBAL
    ambito_padre: Optional[str] = None
    nivel_anidamiento: int = 0

    # Información de ubicación
    linea_declaracion: int = 0
    columna_declaracion: int = 0
    archivo: str = ""

    # Información de memoria (simulada)
    direccion_memoria: str = ""
    offset_pila: int = 0
    tamaño_bytes: int = 0

    # Información de uso
    veces_usado: int = 0
    lineas_uso: List[int] = field(default_factory=list)
    modificado: bool = False
    ultima_modificacion: str = ""

    # Metadatos adicionales
    docstring: Optional[str] = None
    anotacion_tipo: Optional[str] = None
    dependencias: Set[str] = field(default_factory=set)
    etiquetas: Set[str] = field(default_factory=set)

    def __post_init__(self):
        """Inicializa valores por defecto después de la creación"""
        if not self.direccion_memoria:
            self.direccion_memoria = self._generar_direccion()
        if self.tipo_dato != TipoDato.DESCONOCIDO:
            self.tamaño_bytes = self._calcular_tamaño()
        if not self.ultima_modificacion:
            self.ultima_modificacion = datetime.now().isoformat()

    def _generar_direccion(self) -> str:
        """Genera una dirección de memoria simulada"""
        hash_obj = hashlib.md5(f"{self.nombre}{datetime.now().timestamp()}".encode())
        return f"0x{hash_obj.hexdigest()[:8]}"

    def _calcular_tamaño(self) -> int:
        """Calcula el tamaño en bytes del tipo de dato"""
        tamaños = {
            TipoDato.ENTERO: 4,
            TipoDato.FLOTANTE: 8,
            TipoDato.CADENA: 24,  # Aproximado
            TipoDato.BOOLEANO: 1,
            TipoDato.LISTA: 56,   # Aproximado
            TipoDato.DICCIONARIO: 64,  # Aproximado
            TipoDato.TUPLA: 40,   # Aproximado
            TipoDato.CONJUNTO: 48,  # Aproximado
            TipoDato.NINGUNO: 0,
            TipoDato.DESCONOCIDO: 8
        }
        return tamaños.get(self.tipo_dato, 8)

    def registrar_uso(self, linea: int):
        """Registra un uso del símbolo"""
        self.veces_usado += 1
        if linea not in self.lineas_uso:
            self.lineas_uso.append(linea)

    def actualizar_valor(self, nuevo_valor: Any, linea: int):
        """Actualiza el valor del símbolo"""
        self.valor = nuevo_valor
        self.modificado = True
        self.ultima_modificacion = datetime.now().isoformat()
        self.registrar_uso(linea)

    def to_dict(self) -> Dict[str, Any]:
        """Convierte el símbolo a diccionario para visualización"""
        return {
            'nombre': self.nombre,
            'tipo_simbolo': self.tipo_simbolo.value,
            'tipo_dato': self.tipo_dato.value,
            'valor': self.valor,
            'valor_inicial': self.valor_inicial,
            'es_constante': self.es_constante,
            'es_mutable': self.es_mutable,
            'ambito': self.ambito.value,
            'ambito_padre': self.ambito_padre,
            'nivel_anidamiento': self.nivel_anidamiento,
            'linea_declaracion': self.linea_declaracion,
            'columna_declaracion': self.columna_declaracion,
            'direccion_memoria': self.direccion_memoria,
            'offset_pila': self.offset_pila,
            'tamaño_bytes': self.tamaño_bytes,
            'veces_usado': self.veces_usado,
            'lineas_uso': self.lineas_uso,
            'modificado': self.modificado,
            'ultima_modificacion': self.ultima_modificacion,
            'docstring': self.docstring,
            'dependencias': list(self.dependencias),
            'etiquetas': list(self.etiquetas)
        }


# ==========================================
# 2. CLASE PRINCIPAL DE LA TABLA DE SÍMBOLOS
# ==========================================

class TablaSimbolos:
    """
    Tabla de símbolos completa para el análisis de código
    """

    def __init__(self, nombre_programa: str = "main"):
        self.nombre_programa = nombre_programa
        self.simbolos: Dict[str, AtributosSimbolo] = {}
        self.simbolos_por_ambito: Dict[str, List[str]] = {}
        self.ambito_actual = Ambito.GLOBAL
        self.pila_ambitos: List[str] = []
        self.contador_temporal = 0
        self.offset_pila_actual = 0
        self.archivo_actual = ""
        self.linea_actual = 0
        self.columna_actual = 0

        # Inicializar ámbito global
        self._inicializar_ambito("global")

    def _inicializar_ambito(self, nombre_ambito: str):
        """Inicializa un nuevo ámbito"""
        if nombre_ambito not in self.simbolos_por_ambito:
            self.simbolos_por_ambito[nombre_ambito] = []

    def _obtener_ambito_actual(self) -> str:
        """Obtiene el nombre del ámbito actual"""
        if self.pila_ambitos:
            return self.pila_ambitos[-1]
        return "global"

    def _generar_offset(self, tamaño: int = 4) -> int:
        """Genera un offset para la pila"""
        offset = self.offset_pila_actual
        self.offset_pila_actual += tamaño
        return offset

    def declarar_simbolo(self,
                         nombre: str,
                         tipo_simbolo: TipoSimbolo,
                         tipo_dato: TipoDato = TipoDato.DESCONOCIDO,
                         valor: Any = None,
                         es_constante: bool = False,
                         es_mutable: bool = True,
                         ambito: Ambito = None,
                         docstring: Optional[str] = None,
                         anotacion_tipo: Optional[str] = None) -> bool:
        """
        Declara un nuevo símbolo en la tabla

        Returns:
            bool: True si se declaró correctamente, False si ya existe
        """
        if nombre in self.simbolos:
            print(f"⚠️  Advertencia: El símbolo '{nombre}' ya existe")
            return False

        if ambito is None:
            ambito = self.ambito_actual

        ambito_nombre = self._obtener_ambito_actual()

        # Crear atributos del símbolo
        atributos = AtributosSimbolo(
            nombre=nombre,
            tipo_simbolo=tipo_simbolo,
            tipo_dato=tipo_dato,
            valor=valor,
            valor_inicial=valor,
            es_constante=es_constante,
            es_mutable=es_mutable,
            ambito=ambito,
            ambito_padre=ambito_nombre if ambito_nombre != "global" else None,
            nivel_anidamiento=len(self.pila_ambitos),
            linea_declaracion=self.linea_actual,
            columna_declaracion=self.columna_actual,
            archivo=self.archivo_actual,
            offset_pila=self._generar_offset(),
            docstring=docstring,
            anotacion_tipo=anotacion_tipo
        )

        # Registrar el símbolo
        self.simbolos[nombre] = atributos

        # Registrar en el ámbito
        self._inicializar_ambito(ambito_nombre)
        if nombre not in self.simbolos_por_ambito[ambito_nombre]:
            self.simbolos_por_ambito[ambito_nombre].append(nombre)

        print(f"✅ Símbolo declarado: {nombre} ({tipo_simbolo.value}) en {ambito_nombre}")
        return True

    def obtener_simbolo(self, nombre: str) -> Optional[AtributosSimbolo]:
        """Obtiene un símbolo por su nombre"""
        return self.simbolos.get(nombre)

    def actualizar_valor(self, nombre: str, nuevo_valor: Any, linea: int = None):
        """Actualiza el valor de un símbolo existente"""
        if nombre in self.simbolos:
            if linea is None:
                linea = self.linea_actual
            self.simbolos[nombre].actualizar_valor(nuevo_valor, linea)
            print(f"↻ Actualizado: {nombre} = {nuevo_valor}")
        else:
            print(f"❌ Error: Símbolo '{nombre}' no encontrado")

    def registrar_uso(self, nombre: str, linea: int = None):
        """Registra el uso de un símbolo"""
        if nombre in self.simbolos:
            if linea is None:
                linea = self.linea_actual
            self.simbolos[nombre].registrar_uso(linea)

    def entrar_ambito(self, nombre_ambito: str):
        """Entra en un nuevo ámbito"""
        self.pila_ambitos.append(nombre_ambito)
        self._inicializar_ambito(nombre_ambito)
        print(f"📂 Entrando en ámbito: {nombre_ambito}")

    def salir_ambito(self):
        """Sale del ámbito actual"""
        if self.pila_ambitos:
            ambito = self.pila_ambitos.pop()
            print(f"📂 Saliendo del ámbito: {ambito}")

    def obtener_todos_simbolos(self) -> Dict[str, Dict[str, Any]]:
        """Obtiene todos los símbolos como diccionarios"""
        return {nombre: simbolo.to_dict() for nombre, simbolo in self.simbolos.items()}

    def obtener_simbolos_por_ambito(self, ambito: str = None) -> Dict[str, List[str]]:
        """Obtiene los símbolos agrupados por ámbito"""
        if ambito:
            return {ambito: self.simbolos_por_ambito.get(ambito, [])}
        return self.simbolos_por_ambito

    def obtener_estadisticas(self) -> Dict[str, Any]:
        """Obtiene estadísticas de la tabla de símbolos"""
        total = len(self.simbolos)
        por_tipo = {}
        por_ambito = {}

        for simbolo in self.simbolos.values():
            tipo = simbolo.tipo_simbolo.value
            por_tipo[tipo] = por_tipo.get(tipo, 0) + 1

            ambito = simbolo.ambito.value
            por_ambito[ambito] = por_ambito.get(ambito, 0) + 1

        return {
            'total_simbolos': total,
            'por_tipo': por_tipo,
            'por_ambito': por_ambito,
            'total_ambitos': len(self.simbolos_por_ambito),
            'total_offsets_usados': self.offset_pila_actual
        }

    def generar_archivo(self, nombre_archivo: str = "tabla_simbolos.txt"):
        """Genera un archivo con la tabla de símbolos"""
        with open(nombre_archivo, 'w', encoding='utf-8') as f:
            f.write("="*100 + "\n")
            f.write(f"TABLA DE SÍMBOLOS - {self.nombre_programa}\n")
            f.write(f"Generado: {datetime.now().isoformat()}\n")
            f.write("="*100 + "\n\n")

            # Estadísticas
            stats = self.obtener_estadisticas()
            f.write("ESTADÍSTICAS:\n")
            f.write("-"*50 + "\n")
            f.write(f"Total de símbolos: {stats['total_simbolos']}\n")
            f.write(f"Total de ámbitos: {stats['total_ambitos']}\n")
            f.write(f"Offsets usados: {stats['total_offsets_usados']} bytes\n")

            f.write("\nSímbolos por tipo:\n")
            for tipo, count in stats['por_tipo'].items():
                f.write(f"  • {tipo}: {count}\n")

            f.write("\nSímbolos por ámbito:\n")
            for ambito, count in stats['por_ambito'].items():
                f.write(f"  • {ambito}: {count}\n")

            f.write("\n" + "="*100 + "\n")
            f.write("DETALLE DE SÍMBOLOS:\n")
            f.write("="*100 + "\n\n")

            # Detalle de cada símbolo
            for nombre, simbolo in sorted(self.simbolos.items()):
                f.write(f"SÍMBOLO: {nombre}\n")
                f.write("-"*50 + "\n")
                for key, value in simbolo.to_dict().items():
                    f.write(f"  {key:20}: {value}\n")
                f.write("\n")

    def __str__(self) -> str:
        """Representación en string de la tabla"""
        return self._formatear_tabla()

    def _formatear_tabla(self) -> str:
        """Formatea la tabla para visualización"""
        if not self.simbolos:
            return "Tabla de símbolos vacía"

        # Preparar datos para tabla
        headers = ['Nombre', 'Tipo', 'Dato', 'Valor', 'Ámbito', 'Memoria', 'Usos']
        filas = []

        for nombre, simbolo in self.simbolos.items():
            filas.append([
                nombre,
                simbolo.tipo_simbolo.value[:8],
                simbolo.tipo_dato.value[:8],
                str(simbolo.valor)[:12],
                simbolo.ambito.value[:8],
                simbolo.direccion_memoria[:8],
                str(simbolo.veces_usado)
            ])

        # Calcular anchos
        anchos = [len(h) for h in headers]
        for fila in filas:
            for i, celda in enumerate(fila):
                anchos[i] = max(anchos[i], len(celda))

        # Construir tabla
        linea = "+" + "+".join(["-" * (ancho + 2) for ancho in anchos]) + "+"
        resultado = [linea]

        # Encabezados
        resultado.append("|" + "|".join([f" {headers[i]:^{anchos[i]}} " for i in range(len(headers))]) + "|")
        resultado.append(linea)

        # Filas
        for fila in filas:
            resultado.append("|" + "|".join([f" {fila[i]:<{anchos[i]}} " for i in range(len(fila))]) + "|")

        resultado.append(linea)

        return "\n".join(resultado)


# ==========================================
# 3. CONSTRUCTOR DE TABLA DE SÍMBOLOS DESDE CÓDIGO
# ==========================================

class ConstructorTablaSimbolos:
    """
    Clase que construye la tabla de símbolos a partir del código
    """

    def __init__(self):
        self.tabla = TablaSimbolos()
        self.linea_actual = 1
        self.columna_actual = 1

    def construir_desde_archivo(self, archivo: str) -> TablaSimbolos:
        """Construye la tabla de símbolos desde un archivo"""
        with open(archivo, 'r', encoding='utf-8') as f:
            codigo = f.read()

        self.tabla.archivo_actual = archivo
        return self.construir_desde_cadena(codigo)

    def construir_desde_cadena(self, codigo: str) -> TablaSimbolos:
        """Construye la tabla de símbolos desde una cadena de código"""
        try:
            arbol = ast.parse(codigo)
            self._procesar_nodo(arbol)
        except SyntaxError as e:
            print(f"❌ Error de sintaxis: {e}")

        return self.tabla

    def _procesar_nodo(self, nodo, ambito_actual: str = "global"):
        """Procesa recursivamente un nodo del AST"""

        self.linea_actual = getattr(nodo, 'lineno', self.linea_actual)
        self.columna_actual = getattr(nodo, 'col_offset', self.columna_actual)

        # Procesar según el tipo de nodo
        if isinstance(nodo, ast.Module):
            for item in nodo.body:
                self._procesar_nodo(item, ambito_actual)

        elif isinstance(nodo, ast.FunctionDef):
            self._procesar_funcion(nodo, ambito_actual)

        elif isinstance(nodo, ast.Assign):
            self._procesar_asignacion(nodo, ambito_actual)

        elif isinstance(nodo, ast.ClassDef):
            self._procesar_clase(nodo, ambito_actual)

        elif isinstance(nodo, ast.For) or isinstance(nodo, ast.While):
            self._procesar_bucle(nodo, ambito_actual)

        elif isinstance(nodo, ast.If):
            # Procesar el cuerpo del if
            for item in nodo.body:
                self._procesar_nodo(item, ambito_actual)
            for item in nodo.orelse:
                self._procesar_nodo(item, ambito_actual)

        elif isinstance(nodo, ast.Import) or isinstance(nodo, ast.ImportFrom):
            self._procesar_import(nodo)

    def _procesar_funcion(self, nodo: ast.FunctionDef, ambito_actual: str):
        """Procesa una definición de función"""

        # Determinar ámbito
        if ambito_actual == "global":
            nuevo_ambito = f"func_{nodo.name}"
        else:
            nuevo_ambito = f"{ambito_actual}_{nodo.name}"

        # Registrar la función
        tipo_retorno = self._inferir_tipo_retorno(nodo)
        self.tabla.linea_actual = nodo.lineno
        self.tabla.columna_actual = nodo.col_offset

        self.tabla.declarar_simbolo(
            nombre=nodo.name,
            tipo_simbolo=TipoSimbolo.FUNCION,
            tipo_dato=tipo_retorno,
            docstring=ast.get_docstring(nodo),
            ambito=Ambito.GLOBAL if ambito_actual == "global" else Ambito.LOCAL
        )

        # Entrar en el ámbito de la función
        self.tabla.entrar_ambito(nuevo_ambito)

        # Registrar parámetros
        for arg in nodo.args.args:
            nombre_param = arg.arg
            tipo_param = self._inferir_tipo_parametro(arg)

            # Verificar si tiene valor por defecto
            tiene_default = False
            if nodo.args.defaults:
                idx = len(nodo.args.args) - len(nodo.args.defaults)
                if nodo.args.args.index(arg) >= idx:
                    tiene_default = True

            self.tabla.declarar_simbolo(
                nombre=nombre_param,
                tipo_simbolo=TipoSimbolo.PARAMETRO,
                tipo_dato=tipo_param,
                valor=None if not tiene_default else "default",
                ambito=Ambito.LOCAL
            )

        # Procesar el cuerpo de la función
        for item in nodo.body:
            self._procesar_nodo(item, nuevo_ambito)

        # Salir del ámbito de la función
        self.tabla.salir_ambito()

    def _procesar_asignacion(self, nodo: ast.Assign, ambito_actual: str):
        """Procesa una asignación de variable"""

        for target in nodo.targets:
            if isinstance(target, ast.Name):
                nombre = target.id

                # Inferir tipo y valor
                tipo = self._inferir_tipo_expresion(nodo.value)
                valor = self._obtener_valor_constante(nodo.value)

                # Determinar si es declaración o asignación
                if nombre not in self.tabla.simbolos:
                    # Es una declaración
                    self.tabla.linea_actual = nodo.lineno
                    self.tabla.columna_actual = nodo.col_offset

                    self.tabla.declarar_simbolo(
                        nombre=nombre,
                        tipo_simbolo=TipoSimbolo.VARIABLE,
                        tipo_dato=tipo,
                        valor=valor,
                        es_mutable=True,
                        ambito=Ambito.LOCAL if ambito_actual != "global" else Ambito.GLOBAL
                    )
                else:
                    # Es una reasignación
                    self.tabla.actualizar_valor(nombre, valor, nodo.lineno)

    def _procesar_clase(self, nodo: ast.ClassDef, ambito_actual: str):
        """Procesa una definición de clase"""

        self.tabla.declarar_simbolo(
            nombre=nodo.name,
            tipo_simbolo=TipoSimbolo.CLASE,
            tipo_dato=TipoDato.DESCONOCIDO,
            docstring=ast.get_docstring(nodo),
            ambito=Ambito.GLOBAL if ambito_actual == "global" else Ambito.LOCAL
        )

        # Entrar en el ámbito de la clase
        nuevo_ambito = f"class_{nodo.name}"
        self.tabla.entrar_ambito(nuevo_ambito)

        # Procesar el cuerpo de la clase
        for item in nodo.body:
            self._procesar_nodo(item, nuevo_ambito)

        self.tabla.salir_ambito()

    def _procesar_bucle(self, nodo, ambito_actual: str):
        """Procesa un bucle (for/while)"""

        # Procesar el cuerpo del bucle
        for item in nodo.body:
            self._procesar_nodo(item, ambito_actual)

        # Procesar el else del bucle si existe
        if nodo.orelse:
            for item in nodo.orelse:
                self._procesar_nodo(item, ambito_actual)

    def _procesar_import(self, nodo):
        """Procesa una declaración de import"""

        for nombre in nodo.names:
            self.tabla.declarar_simbolo(
                nombre=nombre.name,
                tipo_simbolo=TipoSimbolo.IMPORT,
                tipo_dato=TipoDato.DESCONOCIDO,
                ambito=Ambito.GLOBAL
            )

    def _inferir_tipo_expresion(self, nodo) -> TipoDato:
        """Infiere el tipo de una expresión"""

        if isinstance(nodo, ast.Constant):
            if isinstance(nodo.value, int):
                return TipoDato.ENTERO
            elif isinstance(nodo.value, float):
                return TipoDato.FLOTANTE
            elif isinstance(nodo.value, str):
                return TipoDato.CADENA
            elif isinstance(nodo.value, bool):
                return TipoDato.BOOLEANO
            elif nodo.value is None:
                return TipoDato.NINGUNO

        elif isinstance(nodo, ast.List):
            return TipoDato.LISTA
        elif isinstance(nodo, ast.Dict):
            return TipoDato.DICCIONARIO
        elif isinstance(nodo, ast.Tuple):
            return TipoDato.TUPLA
        elif isinstance(nodo, ast.Set):
            return TipoDato.CONJUNTO
        elif isinstance(nodo, ast.Name):
            # Buscar en la tabla
            simbolo = self.tabla.obtener_simbolo(nodo.id)
            if simbolo:
                return simbolo.tipo_dato

        elif isinstance(nodo, ast.BinOp):
            # Inferir tipo de operación binaria
            tipo_izq = self._inferir_tipo_expresion(nodo.left)
            tipo_der = self._inferir_tipo_expresion(nodo.right)
            if tipo_izq == TipoDato.ENTERO and tipo_der == TipoDato.ENTERO:
                return TipoDato.ENTERO
            elif tipo_izq in [TipoDato.ENTERO, TipoDato.FLOTANTE] and tipo_der in [TipoDato.ENTERO, TipoDato.FLOTANTE]:
                return TipoDato.FLOTANTE
            elif isinstance(nodo.op, ast.Add) and tipo_izq == TipoDato.CADENA:
                return TipoDato.CADENA

        elif isinstance(nodo, ast.Call):
            # Llamada a función
            if isinstance(nodo.func, ast.Name):
                func = self.tabla.obtener_simbolo(nodo.func.id)
                if func:
                    return func.tipo_dato

        return TipoDato.DESCONOCIDO

    def _inferir_tipo_retorno(self, nodo: ast.FunctionDef) -> TipoDato:
        """Infiere el tipo de retorno de una función"""

        # Buscar declaraciones de retorno
        for item in ast.walk(nodo):
            if isinstance(item, ast.Return):
                if item.value:
                    return self._inferir_tipo_expresion(item.value)

        # Si no hay retorno, es None
        return TipoDato.NINGUNO

    def _inferir_tipo_parametro(self, arg: ast.arg) -> TipoDato:
        """Infiere el tipo de un parámetro"""

        # Si tiene anotación de tipo
        if arg.annotation:
            if isinstance(arg.annotation, ast.Name):
                nombre_tipo = arg.annotation.id
                if nombre_tipo == 'int':
                    return TipoDato.ENTERO
                elif nombre_tipo == 'float':
                    return TipoDato.FLOTANTE
                elif nombre_tipo == 'str':
                    return TipoDato.CADENA
                elif nombre_tipo == 'bool':
                    return TipoDato.BOOLEANO
                elif nombre_tipo == 'list':
                    return TipoDato.LISTA
                elif nombre_tipo == 'dict':
                    return TipoDato.DICCIONARIO

        return TipoDato.DESCONOCIDO

    def _obtener_valor_constante(self, nodo) -> Any:
        """Obtiene el valor de una constante"""

        if isinstance(nodo, ast.Constant):
            return nodo.value
        elif isinstance(nodo, ast.List):
            return [self._obtener_valor_constante(item) for item in nodo.elts]
        elif isinstance(nodo, ast.Dict):
            return {self._obtener_valor_constante(k): self._obtener_valor_constante(v)
                    for k, v in zip(nodo.keys, nodo.values)}
        elif isinstance(nodo, ast.Name):
            # Es una referencia a otra variable
            simbolo = self.tabla.obtener_simbolo(nodo.id)
            if simbolo:
                return simbolo.valor
        elif isinstance(nodo, ast.BinOp):
            # Evaluar expresión constante (simplificado)
            izq = self._obtener_valor_constante(nodo.left)
            der = self._obtener_valor_constante(nodo.right)
            if isinstance(nodo.op, ast.Add):
                return izq + der if izq is not None and der is not None else None
            elif isinstance(nodo.op, ast.Sub):
                return izq - der if izq is not None and der is not None else None
            elif isinstance(nodo.op, ast.Mult):
                return izq * der if izq is not None and der is not None else None
            elif isinstance(nodo.op, ast.Div):
                return izq / der if izq is not None and der is not None else None

        return None
````

Valores: True (modificado), False (no modificado)
